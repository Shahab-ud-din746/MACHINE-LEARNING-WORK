```python
# Training data (EnjoySport example)
data = [
    ['Sunny', 'Warm', 'Normal', 'Strong', 'Warm', 'Same',   'Yes'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Warm', 'Same',   'Yes'],
    ['Rainy', 'Cold', 'High',   'Strong', 'Warm', 'Change', 'No'],
    ['Sunny', 'Warm', 'High',   'Strong', 'Cool', 'Change', 'Yes'],
]

def find_s(data):
    hypothesis = None

    for i, row in enumerate(data):
        attributes = row[:-1]   # features
        label = row[-1]         # Yes / No

        if label == 'Yes':
            if hypothesis is None:
                hypothesis = attributes.copy()   # pehla positive example
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != attributes[j]:
                        hypothesis[j] = '?'
            print(f"Example {i+1} ", hypothesis)

    return hypothesis

final_hypothesis = find_s(data)
print("\nFinal Hypothesis:", final_hypothesis)
```
