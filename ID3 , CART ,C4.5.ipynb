<div class="dt-grid">
  <div class="dt-card">
    <h3>🌳 ID3</h3>
    <div class="sub">Iterative Dichotomiser 3 · Quinlan, 1986</div>
    <ul>
      <li>Splits using <b>Information Gain</b></li>
      <li>Works with <b>categorical</b> attributes only</li>
      <li>Creates <b>multi-way</b> splits</li>
      <li>No pruning, so it can overfit</li>
      <li>Biased towards attributes with many values</li>
    </ul>
    <div class="formula">Entropy = −Σ pᵢ log₂(pᵢ)<br>Gain = H(S) − Σ (|Sᵥ|/|S|) · H(Sᵥ)</div>
  </div>
  <div class="dt-card c45">
    <h3>🌲 C4.5</h3>
    <div class="sub">Successor of ID3 · Quinlan, 1993</div>
    <ul>
      <li>Splits using <b>Gain Ratio</b></li>
      <li>Handles <b>continuous</b> and categorical data</li>
      <li>Handles <b>missing values</b></li>
      <li>Supports <b>post-pruning</b></li>
      <li>Fixes ID3's bias towards many-valued attributes</li>
    </ul>
    <div class="formula">SplitInfo = −Σ (|Sᵥ|/|S|) log₂(|Sᵥ|/|S|)<br>GainRatio = Gain / SplitInfo</div>
  </div>
  <div class="dt-card cart">
    <h3>🌴 CART</h3>
    <div class="sub">Classification And Regression Trees · Breiman, 1984</div>
    <ul>
      <li>Splits using <b>Gini Index</b></li>
      <li>Always builds <b>binary</b> trees</li>
      <li>Works for <b>classification and regression</b></li>
      <li>Uses cost-complexity pruning</li>
      <li>Handles numeric and categorical data</li>
    </ul>
    <div class="formula">Gini = 1 − Σ pᵢ²<br>Weighted Gini = Σ (|Sᵥ|/|S|) · Gini(Sᵥ)</div>
  </div>
</div>

```python
import math
import time
from collections import Counter
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. Dataset (PlayTennis)
# ------------------------------------------------------------
df = pd.DataFrame([
    ['Sunny','Hot','High','Weak','No'],
    ['Sunny','Hot','High','Strong','No'],
    ['Overcast','Hot','High','Weak','Yes'],
    ['Rain','Mild','High','Weak','Yes'],
    ['Rain','Cool','Normal','Weak','Yes'],
    ['Rain','Cool','Normal','Strong','No'],
    ['Overcast','Cool','Normal','Strong','Yes'],
    ['Sunny','Mild','High','Weak','No'],
    ['Sunny','Cool','Normal','Weak','Yes'],
    ['Rain','Mild','Normal','Weak','Yes'],
    ['Sunny','Mild','Normal','Strong','Yes'],
    ['Overcast','Mild','High','Strong','Yes'],
    ['Overcast','Hot','Normal','Weak','Yes'],
    ['Rain','Mild','High','Strong','No'],
], columns=['Outlook','Temperature','Humidity','Wind','PlayTennis'])

FEATURES = list(df.columns[:-1])
rows   = df[FEATURES].to_dict('records')
labels = df['PlayTennis'].tolist()

# ------------------------------------------------------------
# 2. Impurity measures
# ------------------------------------------------------------
def entropy(y):
    n = len(y)
    return -sum((c/n) * math.log2(c/n) for c in Counter(y).values())

def gini(y):
    n = len(y)
    return 1 - sum((c/n) ** 2 for c in Counter(y).values())

def group_by(rows, y, f):
    groups = {}
    for r, l in zip(rows, y):
        groups.setdefault(r[f], []).append(l)
    return groups

def info_gain(rows, y, f):
    n = len(y)
    rem = sum(len(g)/n * entropy(g) for g in group_by(rows, y, f).values())
    return entropy(y) - rem

def split_info(rows, y, f):
    n = len(y)
    return -sum(len(g)/n * math.log2(len(g)/n) for g in group_by(rows, y, f).values())

def gain_ratio(rows, y, f):
    si = split_info(rows, y, f)
    return info_gain(rows, y, f) / si if si > 0 else 0.0

def best_gini_split(rows, y, f):
    """Best binary split (feature == value vs. rest) for CART."""
    n, best = len(y), None
    for v in set(r[f] for r in rows):
        left  = [l for r, l in zip(rows, y) if r[f] == v]
        right = [l for r, l in zip(rows, y) if r[f] != v]
        if not left or not right:
            continue
        g = len(left)/n * gini(left) + len(right)/n * gini(right)
        if best is None or g < best[0]:
            best = (g, v)
    return best  # (weighted gini, value) or None

# ------------------------------------------------------------
# 3. Decision tree (ID3 / C4.5 / CART)
# ------------------------------------------------------------
class Node:
    def __init__(self, label=None, feature=None, value=None, majority=None):
        self.label, self.feature, self.value, self.majority = label, feature, value, majority
        self.children = {}

class DecisionTree:
    def __init__(self, criterion='id3'):
        self.criterion = criterion      # 'id3', 'c45' or 'cart'
        self.root = None

    def fit(self, rows, y, features):
        self.root = self._build(rows, y, features)
        return self

    def _build(self, rows, y, features):
        counts = Counter(y)
        maj = counts.most_common(1)[0][0]
        if len(counts) == 1 or not features:
            return Node(label=maj)

        if self.criterion == 'cart':
            best = None
            for f in features:
                s = best_gini_split(rows, y, f)
                if s and (best is None or s[0] < best[0]):
                    best = (s[0], f, s[1])
            if best is None:
                return Node(label=maj)
            _, f, v = best
            node = Node(feature=f, value=v, majority=maj)
            yes = [i for i, r in enumerate(rows) if r[f] == v]
            no  = [i for i, r in enumerate(rows) if r[f] != v]
            for key, idx in (('yes', yes), ('no', no)):
                node.children[key] = self._build([rows[i] for i in idx],
                                                 [y[i] for i in idx], features)
            return node

        # ID3 / C4.5: multi-way split
        if self.criterion == 'id3':
            f = max(features, key=lambda a: info_gain(rows, y, a))
        else:  # C4.5: gain ratio among attributes with at least average gain
            gains = {a: info_gain(rows, y, a) for a in features}
            avg = sum(gains.values()) / len(gains)
            cands = [a for a in features if gains[a] >= avg - 1e-12]
            f = max(cands, key=lambda a: gain_ratio(rows, y, a))

        node = Node(feature=f, majority=maj)
        rest = [a for a in features if a != f]
        for v in set(r[f] for r in rows):
            idx = [i for i, r in enumerate(rows) if r[f] == v]
            node.children[v] = self._build([rows[i] for i in idx],
                                           [y[i] for i in idx], rest)
        return node

    def predict_one(self, row):
        node = self.root
        while node.label is None:
            if self.criterion == 'cart':
                key = 'yes' if row[node.feature] == node.value else 'no'
            else:
                key = row[node.feature]
            if key not in node.children:
                return node.majority
            node = node.children[key]
        return node.label

    def predict(self, rows):
        return [self.predict_one(r) for r in rows]

    # --- tree statistics ---
    def depth(self, node=None):
        node = node or self.root
        if node.label is not None:
            return 0
        return 1 + max(self.depth(c) for c in node.children.values())

    def count(self, node=None):
        node = node or self.root
        if node.label is not None:
            return 1, 1                       # (nodes, leaves)
        n, l = 1, 0
        for c in node.children.values():
            cn, cl = self.count(c)
            n, l = n + cn, l + cl
        return n, l

    def show(self, node=None, indent=''):
        node = node or self.root
        if node.label is not None:
            print(f"{indent}→ {node.label}")
            return
        for key, child in node.children.items():
            if self.criterion == 'cart':
                cond = f"{node.feature} == {node.value}" if key == 'yes' else f"{node.feature} != {node.value}"
            else:
                cond = f"{node.feature} = {key}"
            print(f"{indent}[{cond}]")
            self.show(child, indent + '    ')

# ------------------------------------------------------------
# 4. Root-level split scores for each attribute
# ------------------------------------------------------------
scores = pd.DataFrame({
    'Info Gain (ID3)':  [info_gain(rows, labels, f) for f in FEATURES],
    'Gain Ratio (C4.5)': [gain_ratio(rows, labels, f) for f in FEATURES],
    'Best Weighted Gini (CART)': [best_gini_split(rows, labels, f)[0] for f in FEATURES],
}, index=FEATURES).round(4)
print("Split scores at the root node:\n")
print(scores, "\n")

# ------------------------------------------------------------
# 5. Build the trees and print them
# ------------------------------------------------------------
models = {
    'ID3':  DecisionTree('id3'),
    'C4.5': DecisionTree('c45'),
    'CART': DecisionTree('cart'),
}
for name, m in models.items():
    m.fit(rows, labels, FEATURES)
    print(f"{'=' * 12} {name} Tree {'=' * 12}")
    m.show()
    print()

# ------------------------------------------------------------
# 6. Comparison: accuracy, size, speed
# ------------------------------------------------------------
def loocv(criterion):
    correct = 0
    for i in range(len(rows)):
        tr_x = rows[:i] + rows[i+1:]
        tr_y = labels[:i] + labels[i+1:]
        t = DecisionTree(criterion).fit(tr_x, tr_y, FEATURES)
        correct += t.predict_one(rows[i]) == labels[i]
    return correct / len(rows)

results = []
for name, m in models.items():
    crit = m.criterion
    train_acc = sum(p == a for p, a in zip(m.predict(rows), labels)) / len(rows)

    start = time.perf_counter()s
    for _ in range(200):
        DecisionTree(crit).fit(rows, labels, FEATURES)
    fit_ms = (time.perf_counter() - start) / 200 * 1000

    nodes, leaves = m.count()
    results.append({
        'Algorithm': name,
        'Train Accuracy': round(train_acc, 3),
        'LOOCV Accuracy': round(loocv(crit), 3),
        'Depth': m.depth(),
        'Nodes': nodes,
        'Leaves': leaves,
        'Fit Time (ms)': round(fit_ms, 3),
    })

comparison = pd.DataFrame(results).set_index('Algorithm')
print("Comparison of ID3, C4.5 and CART:\n")
print(comparison)

# ------------------------------------------------------------
# 7. Visual comparison
# ------------------------------------------------------------
colors = ['#00d4ff', '#7b61ff', '#ff7a59']
fig, axes = plt.subplots(1, 4, figsize=(18, 4))
for ax, col in zip(axes, ['LOOCV Accuracy', 'Depth', 'Nodes', 'Fit Time (ms)']):
    bars = ax.bar(comparison.index, comparison[col], color=colors)
    ax.set_title(col, fontweight='bold')
    ax.bar_label(bars, padding=2)
    ax.spines[['top', 'right']].set_visible(False)
plt.suptitle('ID3 vs C4.5 vs CART', fontsize=15, fontweight='bold')
plt.tight_layout()
plt.show()
```
