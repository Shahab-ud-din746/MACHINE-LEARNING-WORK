```python

# ==========================================================
# Algorithm : Bayesian Belief Network
# Project   : Smart Home Lighting Prediction System
# ==========================================================

from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination


# STEP 1: Create the Bayesian Network
model = DiscreteBayesianNetwork([
    ("Occupied", "LightOn"),
    ("LowLight", "LightOn")
])


# STEP 2: Prior probability of room occupancy
occupied_cpd = TabularCPD(
    variable="Occupied",
    variable_card=2,
    values=[
        [0.4],  # Occupied = No
        [0.6]   # Occupied = Yes
    ],
    state_names={"Occupied": ["No", "Yes"]}
)


# STEP 3: Prior probability of low light
low_light_cpd = TabularCPD(
    variable="LowLight",
    variable_card=2,
    values=[
        [0.3],  # LowLight = No
        [0.7]   # LowLight = Yes
    ],
    state_names={"LowLight": ["No", "Yes"]}
)


# STEP 4: Conditional probability of the light
# Evidence order: Occupied, LowLight
# Columns:
# (No, No), (No, Yes), (Yes, No), (Yes, Yes)

light_on_cpd = TabularCPD(
    variable="LightOn",
    variable_card=2,
    values=[
        [0.99, 0.70, 0.90, 0.05],  # LightOn = No
        [0.01, 0.30, 0.10, 0.95]   # LightOn = Yes
    ],
    evidence=["Occupied", "LowLight"],
    evidence_card=[2, 2],
    state_names={
        "LightOn": ["No", "Yes"],
        "Occupied": ["No", "Yes"],
        "LowLight": ["No", "Yes"]
    }
)


# STEP 5: Add CPDs and validate the model
model.add_cpds(
    occupied_cpd,
    low_light_cpd,
    light_on_cpd
)

print("💡 BAYESIAN BELIEF NETWORK")
print("SMART HOME LIGHTING PREDICTION")
print("=" * 40)

if not model.check_model():
    raise ValueError("Bayesian Network model validation failed.")

print("Model validation: SUCCESS")


# STEP 6: Perform probabilistic inference
inference = VariableElimination(model)

result = inference.query(
    variables=["LightOn"],
    evidence={
        "Occupied": "Yes",
        "LowLight": "Yes"
    },
    show_progress=False
)


# STEP 7: Extract probabilities
probability_on = result.get_value(LightOn="Yes") * 100
probability_off = result.get_value(LightOn="No") * 100


# STEP 8: Display results
print("\n📊 SENSOR READINGS")
print("-" * 40)
print("Room occupied : YES")
print("Low light      : YES")

print("\n✨ PREDICTION RESULTS")
print("-" * 40)
print(f"Light ON probability  : {probability_on:.1f}%")
print(f"Light OFF probability : {probability_off:.1f}%")

if probability_on >= 50:
    print("\n💡 Recommendation: TURN THE LIGHT ON")
else:
    print("\n🌙 Recommendation: KEEP THE LIGHT OFF")

print("\n✅ Prediction completed successfully.")
```
